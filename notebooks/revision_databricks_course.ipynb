# Change Data Feed Learning

```python
%sql
CREATE TABLE yatindra_org.microfin.change_data_feed_practice (
id INT,
name STRING,
amount DOUBLE
);

INSERT INTO yatindra_org.microfin.change_data_feed_practice VALUES (1, 'alice', 100.0), (2, 'bob', 200.0), (3, 'carol', 300.0);

ALTER TABLE yatindra_org.microfin.change_data_feed_practice SET TBLPROPERTIES (delta.enablechangedatafeed = true);
```

```python
%sql
show create table yatindra_org.microfin.change_data_feed_practice
```

```python
%sql
-- Update
UPDATE yatindra_org.microfin.change_data_feed_practice SET amount = amount * 1.1 WHERE id = 2;
-- Insert additional
INSERT INTO yatindra_org.microfin.change_data_feed_practice VALUES (4, 'dan', 400.0);
-- delete
DELETE FROM yatindra_org.microfin.change_data_feed_practice WHERE id = 1;
```
